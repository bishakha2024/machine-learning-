# Simple Thresholding

```python
import cv2 as cv
import numpy as np
from matplotlib import pyplot as plt

# Load the image in grayscale
image_path = 'cb.jpeg'  
img = cv.imread(image_path, cv.IMREAD_GRAYSCALE)

# Check if the image is loaded correctly
assert img is not None, "file could not be read, check with os.path.exists()"

# Apply different thresholding techniques
ret, thresh1 = cv.threshold(img, 127, 255, cv.THRESH_BINARY)
ret, thresh2 = cv.threshold(img, 127, 255, cv.THRESH_BINARY_INV)
ret, thresh3 = cv.threshold(img, 127, 255, cv.THRESH_TRUNC)
ret, thresh4 = cv.threshold(img, 127, 255, cv.THRESH_TOZERO)
ret, thresh5 = cv.threshold(img, 127, 255, cv.THRESH_TOZERO_INV)

# Define titles for each technique
titles = ['Original Image', 'BINARY', 'BINARY_INV', 'TRUNC', 'TOZERO', 'TOZERO_INV']
images = [img, thresh1, thresh2, thresh3, thresh4, thresh5]

# Display the images using matplotlib
plt.figure(figsize=(10, 6))  # Adjust the figure size for better layout
for i in range(6):
    plt.subplot(2, 3, i + 1)
    plt.imshow(images[i], cmap='gray', vmin=0, vmax=255)
    plt.title(titles[i])
    plt.xticks([]), plt.yticks([])

# Show the plots
plt.tight_layout()
plt.show()
```

# know basics-
BINARY: Converts all pixel values above the threshold to 255 (white), and below the threshold to 0 (black).
BINARY_INV: Inverts the BINARY thresholding, i.e., all values above the threshold become 0 (black), and below become 255 (white).
TRUNC: Truncates pixel values that are above the threshold to the threshold value (i.e., pixel values above the threshold are set to 127).
TOZERO: All pixels below the threshold are set to 0 (black), and values above the threshold remain unchanged.
TOZERO_INV: Inverse of TOZERO, i.e., all pixels above the threshold are set to 0, and values below the threshold remain unchanged.

# Adaptive Thresholding

```python

# Apply median blur to reduce noise
img = cv.medianBlur(img, 5)

# Apply different thresholding techniques
ret, th1 = cv.threshold(img, 127, 255, cv.THRESH_BINARY)  # Global Thresholding
th2 = cv.adaptiveThreshold(img, 255, cv.ADAPTIVE_THRESH_MEAN_C,  # Adaptive Thresholding (Mean)
                           cv.THRESH_BINARY, 11, 2)
th3 = cv.adaptiveThreshold(img, 255, cv.ADAPTIVE_THRESH_GAUSSIAN_C,  # Adaptive Thresholding (Gaussian)
                           cv.THRESH_BINARY, 11, 2)

# Titles for each technique
titles = ['Original Image', 'Global Thresholding (v = 127)',
          'Adaptive Mean Thresholding', 'Adaptive Gaussian Thresholding']

# Images to display
images = [img, th1, th2, th3]

# Display the results using matplotlib
plt.figure(figsize=(10, 6))
for i in range(4):
    plt.subplot(2, 2, i + 1)
    plt.imshow(images[i], cmap='gray')
    plt.title(titles[i])
    plt.xticks([]), plt.yticks([])

# Show the plots
plt.tight_layout()
plt.show()
```

# basics-
* Global Thresholding (v = 127): A global thresholding with a fixed value of 127.
* Adaptive Thresholding (Mean): An adaptive thresholding where the threshold is computed based on the local neighborhood using the mean of the area.
* Adaptive Thresholding (Gaussian): An adaptive thresholding where the threshold is computed based on the local neighborhood using a Gaussian weighted sum of the area.


# Otsu's Binarization

```python

# Apply global thresholding
ret1, th1 = cv.threshold(img, 127, 255, cv.THRESH_BINARY)

# Apply Otsu's thresholding
ret2, th2 = cv.threshold(img, 0, 255, cv.THRESH_BINARY + cv.THRESH_OTSU)

# Apply Gaussian filtering followed by Otsu's thresholding
blur = cv.GaussianBlur(img, (5, 5), 0)
ret3, th3 = cv.threshold(blur, 0, 255, cv.THRESH_BINARY + cv.THRESH_OTSU)

# Prepare images and titles for displaying
images = [img, 0, th1, 
          img, 0, th2, 
          blur, 0, th3]
titles = ['Original Noisy Image', 'Histogram', 'Global Thresholding (v=127)',
          'Original Noisy Image', 'Histogram', "Otsu's Thresholding",
          'Gaussian Filtered Image', 'Histogram', "Otsu's Thresholding"]

# Create subplots for the images and their histograms
plt.figure(figsize=(12, 10))  # Adjust figure size for better layout
for i in range(3):
    # Original Image
    plt.subplot(3, 3, i * 3 + 1)
    plt.imshow(images[i * 3], cmap='gray')
    plt.title(titles[i * 3])
    plt.xticks([]), plt.yticks([])

    # Histogram of the original image
    plt.subplot(3, 3, i * 3 + 2)
    plt.hist(images[i * 3].ravel(), 256)
    plt.title(titles[i * 3 + 1])
    plt.xticks([]), plt.yticks([])

    # Thresholded Image
    plt.subplot(3, 3, i * 3 + 3)
    plt.imshow(images[i * 3 + 2], cmap='gray')
    plt.title(titles[i * 3 + 2])
    plt.xticks([]), plt.yticks([])

# Show the plots
plt.tight_layout()
plt.show()
```
